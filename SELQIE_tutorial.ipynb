# SELQIE Tutorial

Basic tutorial for SELQIE

First, import packages

```python
import numpy as np
import pyscf
import SELQIE as slq
```

Check that SELQIE is installed

```python
slq.config.hello()
```

Setup system in PySCF

```python
mol = pyscf.gto.Mole()
mol.verbose = 0
mol.atom = """
O   0.000000   0.000000   0.000000
H   0.000000   0.758602   0.504284
H   0.000000  -0.758602   0.504284
"""
mol.basis = 'sto-3g'
mol.symmetry = "c2v"
mol.build()
```

Create SELQIE `settings` object, in this tutorial we will be using HCI


*   Method - "sci", "cisd", or "fci"
*   Sel_method - "hci"
*   n_up - number of up electrons
*   n_down - number of down electrons
*   conv - energy convergence target
*   cmin - selection pruning parameter
*   hci_epsilon - HCI accuracy parameter


```python
n_alpha, n_beta = mol.nelec
hci_settings = slq.config.settings(method="sci",sel_method="hci",n_up=n_alpha,n_down=n_beta,conv=1e-6,cmin=1e-6, hci_epsilon=0.001)
```

Now, run the calculation. Results are outputted into a `state` object

```python
hci_state = slq.selected_ci.solve(mol, hci_settings)
print(f"HCI Energy: {hci_state.energy} Ha")
```

The determinant basis and CI vector can also be extracted

```python
print(f"Number of determinants used: {hci_state.basis.shape[1]}")
print(f"Number of coefficients in vec: {len(hci_state.civec)}")
print(f"Norm of CI vector: {np.linalg.norm(hci_state.civec)}")
```
